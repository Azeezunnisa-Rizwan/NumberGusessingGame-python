## Loops and Conditions

```python
# 1. while loop
i = 1
while i <= 5:
    print(i)
    i += 1

count = 0
while count < 3:
    print("Hello")
    count += 1


# 2. for loop
for i in range(5):
    print(i)

for ch in "Azeez":
    print(ch)


# 3. if else
age = 20
if age >= 18:
    print("Adult")
else:
    print("Minor")

num = -5
if num >= 0:
    print("Positive")
else:
    print("Negative")


# 4. if elif else
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")

num = 0
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")


# 5. user defined functions without for loop, if loop
def greet():
    print("Hello, welcome!")
greet()

def add(a, b):
    return a + b
print(add(3, 4))


# 6. user defined functions with for loop
def print_numbers(n):
    for i in range(n):
        print(i)
print_numbers(5)

def sum_list(lst):
    total = 0
    for num in lst:
        total += num
    return total
print(sum_list([1, 2, 3, 4]))


# 7. user defined functions with for loop and if condition
def find_evens(n):
    for i in range(n):
        if i % 2 == 0:
            print(i, "is even")
find_evens(10)

def count_vowels(word):
    count = 0
    for ch in word:
        if ch in "aeiouAEIOU":
            count += 1
    return count
print(count_vowels("Azeez"))


# 8. break
for i in range(10):
    if i == 5:
        break
    print(i)

n = 0
while True:
    if n == 3:
        break
    print(n)
    n += 1


# 9. continue
for i in range(5):
    if i == 2:
        continue
    print(i)

n = 0
while n < 5:
    n += 1
    if n == 3:
        continue
    print(n)


# 10. for loop with in operator
for ch in "hello":
    print(ch, "is in hello")

fruits = ["apple", "banana", "mango"]
for fruit in fruits:
    print(fruit)


# 11. for loop with if statement
for i in range(10):
    if i % 2 == 0:
        print(i, "is even")

names = ["Azeez", "John", "Ali"]
for name in names:
    if len(name) > 3:
        print(name)


# 12. for loop with not in operator
vowels = "aeiou"
for ch in "python":
    if ch not in vowels:
        print(ch, "is a consonant")

allowed = ["a", "b", "c"]
for item in ["a", "x", "c", "z"]:
    if item not in allowed:
        print(item, "not allowed")


# 13. if with in operator
word = "hello"
if "e" in word:
    print("'e' found in word")

fruits = ["apple", "banana"]
if "apple" in fruits:
    print("apple is available")


# 14. if with not in operator
word = "hello"
if "z" not in word:
    print("'z' not found in word")

fruits = ["apple", "banana"]
if "mango" not in fruits:
    print("mango is not available")
```

## Task 3 - Rock, Paper, Scissors Game

```python
import random

while True:
    user = input("Enter rock, paper, or scissors: ").lower()
    
    try:
        if user not in ["rock", "paper", "scissors"]:
            raise ValueError("Invalid choice!")
    except ValueError as e:
        print(e)
        continue
    
    computer = random.choice(["rock", "paper", "scissors"])
    print("Computer chose:", computer)
    
    if user == computer:
        print("It's a tie!")
    elif (user == "rock" and computer == "scissors") or \
         (user == "paper" and computer == "rock") or \
         (user == "scissors" and computer == "paper"):
        print("You win!")
    else:
        print("Computer wins!")
    
    again = input("Play again? (yes/no): ").lower()
    if again != "yes":
        break
```


## Word Guessing Game (Mini Hangman)

```python
import random

words = ["python", "machine", "learning", "bridge"]
word = random.choice(words)
guessed = ["_"] * len(word)
attempts = 6

print(" ".join(guessed))

while attempts > 0 and "_" in guessed:
    letter = input("Guess a letter: ").lower()
    
    if letter in word:
        for i in range(len(word)):
            if word[i] == letter:
                guessed[i] = letter
    else:
        attempts -= 1
        print("Wrong! Attempts left:", attempts)
    
    print(" ".join(guessed))

if "_" not in guessed:
    print("You win! The word was:", word)
else:
    print("You lost! The word was:", word)
```
